```python
from itertools import permutations

def solve_cryptarithmetic(puzzle):
    """ Solves the cryptarithmetic puzzle.
    puzzle: a string containing the puzzle in the form of "WORD + WORD = RESULT"
    Returns a dictionary mapping letters to digits, or None if no solution is found.
    """
    words_str = puzzle.replace(' ', '').replace('+', '').replace('=', '')
    words_split = puzzle.split()

    # Extract individual words for calculation later
    term1_str = words_split[0]
    term2_str = words_split[2]
    result_str = words_split[4]

    letters = set(words_str)

    if len(letters) > 10:
        return None  # More than 10 letters, no solution is possible

    for perm in permutations(range(10), len(letters)):
        mapping = dict(zip(letters, perm))

        # Check for leading zeros
        if mapping[term1_str[0]] == 0 or mapping[term2_str[0]] == 0 or mapping[result_str[0]] == 0:
            continue

        # Convert words to numbers
        num1 = sum(mapping[c] * (10 ** (len(term1_str) - i - 1)) for i, c in enumerate(term1_str))
        num2 = sum(mapping[c] * (10 ** (len(term2_str) - i - 1)) for i, c in enumerate(term2_str))
        res_num = sum(mapping[c] * (10 ** (len(result_str) - i - 1)) for i, c in enumerate(result_str))

        if num1 + num2 == res_num:
            return mapping

    return None

# Example usage
puzzle = "SEND + MORE = MONEY"
mapping = solve_cryptarithmetic(puzzle)

if mapping:
    print(mapping)
    translated_puzzle = puzzle
    for letter, digit in mapping.items():
        translated_puzzle = translated_puzzle.replace(letter, str(digit))
    print(translated_puzzle)
else:
    print('No solution found.')
```
