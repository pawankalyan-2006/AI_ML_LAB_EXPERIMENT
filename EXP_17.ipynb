```python
# Python Program to Implement DFS

# Graph represented using a dictionary
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': []
}

# Set to keep track of visited nodes
visited = set()


# DFS function
def dfs_visit(node):
    # Add current node to visited set
    visited.add(node)

    # Display the visited node
    print(node, end=" ")

    # Visit each unvisited neighbor
    for neighbor in graph[node]:
        if neighbor not in visited:
            dfs_visit(neighbor)


# Starting node
start_node = 'A'

# Call DFS
print("DFS Traversal:")
dfs_visit(start_node)
```
