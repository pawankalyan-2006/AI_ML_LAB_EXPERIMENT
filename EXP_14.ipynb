```python
from collections import deque

# (M_left, C_left, Boat)
start = (3, 3, 0)
goal = (0, 0, 1)

moves = [
    (1, 0),
    (0, 1),
    (2, 0),
    (0, 2),
    (1, 1)
]


def valid(state):
    ml, cl, boat = state
    mr = 3 - ml
    cr = 3 - cl

    if ml < 0 or cl < 0 or mr < 0 or cr < 0:
        return False

    if ml > 0 and cl > ml:
        return False

    if mr > 0 and cr > mr:
        return False

    return True


def solve():
    queue = deque([(start, [])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path + [state]

        ml, cl, boat = state

        for m, c in moves:

            if boat == 0:
                new_state = (ml - m, cl - c, 1)
            else:
                new_state = (ml + m, cl + c, 0)

            if valid(new_state) and new_state not in visited:
                visited.add(new_state)
                queue.append(
                    (new_state, path + [state])
                )

    return None


solution = solve()

print("Missionaries and Cannibals Solution")
print("------------------------------------")

for state in solution:
    ml, cl, boat = state
    mr = 3 - ml
    cr = 3 - cl

    side = "Left" if boat == 0 else "Right"

    print(
        f"Left: M={ml}, C={cl} | "
        f"Right: M={mr}, C={cr} | "
        f"Boat: {side}"
    )
```
