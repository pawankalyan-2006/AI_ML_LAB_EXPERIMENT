```python
# Python Program to Implement Travelling Salesman Problem

from itertools import permutations

# Distance matrix
distance = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

# Number of cities
n = len(distance)

# Starting city
start = 0

# List of remaining cities
cities = list(range(1, n))

# Initialize best distance and best route
best_distance = float('inf')
best_route = []

# Generate every possible permutation
for perm in permutations(cities):

    # Create complete route
    route = (start,) + perm + (start,)

    # Calculate total distance
    total_distance = 0

    for i in range(len(route) - 1):
        total_distance += distance[route[i]][route[i + 1]]

    # Update best distance and route
    if total_distance < best_distance:
        best_distance = total_distance
        best_route = route

# Display result
print("Travelling Salesman Problem")
print("---------------------------")

print("Best Route:", end=" ")

for city in best_route:
    print(city + 1, end=" ")

print("\nBest Distance:", best_distance)
```
