```python
# Python Program for Map Coloring using CSP

# Available colors
colors = ['Red', 'Green', 'Blue']

# Map represented as a graph
# Each country is connected to its neighboring countries
neighbors = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}

# Assignment dictionary
assignment = {}


# Check whether assigning a color is consistent
def is_consistent(country, color, assignment):

    for neighbor in neighbors[country]:

        if neighbor in assignment:
            if assignment[neighbor] == color:
                return False

    return True


# Backtracking function
def backtrack(assignment):

    # Check if assignment is complete
    if len(assignment) == len(neighbors):
        return assignment

    # Select an unassigned country
    for country in neighbors:
        if country not in assignment:
            break

    # Try each color
    for color in colors:

        # Check consistency
        if is_consistent(country, color, assignment):

            # Assign color
            assignment[country] = color

            # Recursively call backtrack
            result = backtrack(assignment)

            # If solution is found
            if result is not None:
                return result

            # Remove assignment if it fails
            del assignment[country]

    # Return failure
    return None


# Call backtracking function
solution = backtrack(assignment)


# Display the result
if solution:
    print("Map Coloring using CSP")
    print("----------------------")

    for country, color in solution.items():
        print(country, "->", color)

else:
    print("No solution exists.")
```
