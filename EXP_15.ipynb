```python
from typing import List, Tuple

def get_successors(state: Tuple[Tuple[int, int], Tuple[Tuple[int, ...]]]) -> List[Tuple[Tuple[int, int], Tuple[Tuple[int, ...]]]]:
    """ Generate all possible valid states that can be reached from a given state. """
    position, grid_tuple = state
    successors = []
    for dx, dy in [(0, 1), (0, -1), (1, 0), (-1, 0)]:
        x, y = position
        new_x, new_y = x + dx, y + dy
        if 0 <= new_x < len(grid_tuple) and 0 <= new_y < len(grid_tuple[0]):
            # Convert grid_tuple (tuple of tuples) to a mutable list of lists for modification
            current_grid_list = [list(row) for row in grid_tuple]
            if current_grid_list[new_x][new_y] == 1:
                current_grid_list[new_x][new_y] = 0
            # Convert the modified list of lists back to a tuple of tuples for the successor state
            successor_grid_tuple = tuple(tuple(row) for row in current_grid_list)
            successors.append(((new_x, new_y), successor_grid_tuple))
    return successors

def depth_first_search(state: Tuple[Tuple[int, int], Tuple[Tuple[int, ...]]], visited: set) -> bool:
    """ Find the solution using a depth-first search algorithm. """
    # 'state' is already in a hashable form (position tuple, grid tuple of tuples)
    if all(all(cell == 0 for cell in row) for row in state[1]):
        return True
    visited.add(state)
    for successor in get_successors(state):
        # 'successor' is already in a hashable form as returned by get_successors
        if successor not in visited:
            if depth_first_search(successor, visited):
                return True
    return False

if __name__ == '__main__':
    # Initial grid is a list of lists
    initial_raw_grid = [[0, 1, 1, 1],
                        [0, 0, 1, 0],
                        [1, 0, 0, 1],
                        [1, 1, 1, 0]]
    # Convert initial_raw_grid to a tuple of tuples for the hashable initial_state
    initial_hashable_grid = tuple(tuple(row) for row in initial_raw_grid)
    initial_state = ((0, 0), initial_hashable_grid)

    if depth_first_search(initial_state, set()):
        print("The room can be cleaned.")
    else:
        print("The room cannot be cleaned.")
```
