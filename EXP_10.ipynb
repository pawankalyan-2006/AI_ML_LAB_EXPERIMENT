```python
import heapq

# Calculate Manhattan Distance
def manhattan_distance(state, goal):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            current_row = i // 3
            current_col = i % 3

            goal_index = goal.index(state[i])
            goal_row = goal_index // 3
            goal_col = goal_index % 3

            distance += abs(current_row - goal_row) + abs(current_col - goal_col)

    return distance


# Generate successor states
def get_successors(state):
    successors = []

    zero_index = state.index(0)
    row = zero_index // 3
    col = zero_index % 3

    # Possible movements of empty tile
    moves = [
        (-1, 0),  # Up
        (1, 0),   # Down
        (0, -1),  # Left
        (0, 1)    # Right
    ]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_index = new_row * 3 + new_col

            new_state = list(state)

            # Swap empty tile with neighboring tile
            new_state[zero_index], new_state[new_index] = \
                new_state[new_index], new_state[zero_index]

            successors.append(tuple(new_state))

    return successors


# Print puzzle state
def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print()


# A* algorithm
def solve_puzzle(start, goal):

    # Frontier priority queue
    frontier = []

    # g = cost so far
    # h = Manhattan distance
    # f = g + h
    initial_priority = manhattan_distance(start, goal)

    heapq.heappush(
        frontier,
        (initial_priority, 0, start, [])
    )

    explored = set()

    while frontier:

        # Step 1: Dequeue state with lowest priority
        priority, cost, state, path = heapq.heappop(frontier)

        # Step 2: Check goal state
        if state == goal:
            return path + [state]

        # Step 3: Add state to explored set
        if state in explored:
            continue

        explored.add(state)

        # Step 4: Generate successor states
        for successor in get_successors(state):

            # Step 5: Add unexplored successor to frontier
            if successor not in explored:

                new_cost = cost + 1
                heuristic = manhattan_distance(successor, goal)
                new_priority = new_cost + heuristic

                heapq.heappush(
                    frontier,
                    (
                        new_priority,
                        new_cost,
                        successor,
                        path + [state]
                    )
                )

    return None


# Main program
start_state = (
    1, 2, 3,
    4, 0, 6,
    7, 5, 8
)

goal_state = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

print("Initial State:")
print_state(start_state)

print("Goal State:")
print_state(goal_state)

solution = solve_puzzle(start_state, goal_state)

if solution:
    print("Solution found!")
    print("Number of moves:", len(solution) - 1)
    print()

    for step, state in enumerate(solution):
        print("Step", step)
        print_state(state)
else:
    print("No solution exists.")
```
